# 3. Data Wrangling
Clean the collected data and create the target label `Class` (1 = landed, 0 = did not land).

```python
import pandas as pd
import numpy as np
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
df.head()
```

```python
print(df.isnull().sum() / len(df) * 100)
print(df.dtypes)
```

```python
print(df['LaunchSite'].value_counts())
print(df['Orbit'].value_counts())
landing_outcomes = df['Outcome'].value_counts()
print(landing_outcomes)
```

```python
# Outcomes that mean the first stage did NOT land successfully
bad_outcomes = {'False ASDS', 'False Ocean', 'False RTLS', 'None ASDS', 'None None'}
df['Class'] = df['Outcome'].apply(lambda x: 0 if x in bad_outcomes else 1)
print(df['Class'].value_counts())
print('Overall success rate:', df['Class'].mean())
df.to_csv('../data/dataset_part_2.csv', index=False)
```
